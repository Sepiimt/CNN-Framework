```python
# --- Imports ---
from pathlib import Path
import sys
# --- Resolving Package Path ---
notebook_dir = Path.cwd()
src_path = (notebook_dir / ".." / "src").resolve()
if str(src_path) not in sys.path:
    sys.path.insert(0, str(src_path))
# --- Importing API Files ---
import torch
from svhn.models._backbone import BackBone
from svhn.models._dneck import FeaturePyramidNeck
```

```python
# --- Initialize Backbone ---
backbone = BackBone()


# --- Initialize Neck ---
neck = FeaturePyramidNeck(
    low_channels = 32,
    high_channels = 128,
    out_channels = 64
)


# --- Dummy Input ---
images = torch.randn(3, 3, 32, 32)


# --- Backbone ---
backbone_features = backbone(images)


# --- Neck ---
pyramid = neck(backbone_features)


# --- Inspect Outputs ---
for name, feature in pyramid.items():

    print(name, feature.shape)
```
