## --- Train Data Processing ---

```python
path = '../data/raw/train/'
```

```python
y1, x1, height_ls, width_ls, labels_ls, names_ls = mat_data_extraction(path)
```

```python
y2, x2 = coordinate_to_point(y1, x1, height_ls, width_ls)
del width_ls, height_ls
```

```python
labels_ls[0] = [float(1), float(9)] #> Index 0 is flawed, this is a patch
labels_ls = replace_10_labels_with_0(labels_ls)
```

```python
y1, y2, x1, x2 = inner_data_conversion_to_numpy(y1, y2, x1, x2)
```

```python
random_sample_visulization(
    5, 
    path,
    y1, y2, x1, x2,
    labels_ls, names_ls)
```

```python
save_path = "../data/extracted/annotations/train/"
save_annotation_data(save_path, y1, y2, x1, x2, labels_ls, names_ls)
```
